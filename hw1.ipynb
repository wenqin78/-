根据下列递归公式, 计算下列 $T(n)$ 对应的的渐近上界。要求所求的边界尽可能的紧（tight）, 请写明步骤。

### 1. $T(n) = T(n/4) + 1$

* **分析：** 该递归方程表征了一种单分支的分治策略，符合主定理 (Master Theorem) 的标准范式 $T(n) = aT(n/b) + f(n)$。此处问题规模以 $1/4$ 的比例衰减，且不产生并行的子问题分支。提取系统参数：$a=1, b=4, f(n)=1$。
* **计算：** 评估递归树底层的叶节点状态界限：$n^{\log_b a} = n^{\log_4 1}$。由于任何非零标量的零次幂为 $1$，得 $n^0 = 1$。
* **比较：** 将驱动项（当前层合并代价）$f(n) = 1$ 与叶节点界限 $n^{\log_b a} = 1$ 进行对比。二者渐近等价，满足 $f(n) = \Theta(n^{\log_b a})$。这表明在递归树的各向同性展开中，每一层的渐近时间开销分布是均匀的。
* **结论：** 依据主定理 Case 2（均衡主导情形），系统总时间开销等于单层开销与递归深度的乘积。
  $$T(n) = \Theta(n^{\log_b a} \log n) = \Theta(1 \cdot \log n) = \Theta(\log n)$$
* **渐近上界：** $O(\log n)$

---

### 2. $T(n) = 3T(n/3) + n^2$

* **分析：** 本方程描述了一个高频裂变且合并代价昂贵的分治过程。提取系统参数：$a=3, b=3, f(n)=n^2$。
* **计算：** 计算叶节点状态数界限：$n^{\log_b a} = n^{\log_3 3} = n^1 = n$。
* **比较：** 考察驱动项 $f(n) = n^2$ 与叶节点开销 $n^{\log_b a} = n$。显然存在常数 $\epsilon > 0$（例如取 $\epsilon = 1$），使得 $f(n) = \Omega(n^{\log_b a + \epsilon})$ 成立，即顶层合并代价在多项式意义上严格大于底端开销。进一步验证正则性条件 (Regularity Condition)：需存在常数 $c < 1$，使得对于足够大的 $n$ 有 $af(n/b) \le cf(n)$。代入得 $3(n/3)^2 = \frac{1}{3}n^2 \le cn^2$，取 $c=1/3$ 即满足该条件。这验证了计算负载在递归树中呈现“顶端重载 (Top-heavy)”态势。
* **结论：** 依据主定理 Case 3（根节点主导情形），整体渐近行为被根节点的合并开销所控制。
  $$T(n) = \Theta(f(n)) = \Theta(n^2)$$
* **渐近上界：** $O(n^2)$

### 3. $T(n) = T(2n/3) + 1$

* **分析：** 该方程可通过代数同构转化为标准主定理形式：$T(n) = 1 \cdot T(n / \frac{3}{2}) + 1$。这反映了一个非对称缩减的单分支递归模型。提取参数：$a=1, b=3/2, f(n)=1$。
* **计算：** 评估叶节点规模量级：$n^{\log_b a} = n^{\log_{3/2} 1} = n^0 = 1$。
* **比较：** 对比工作量 $f(n) = 1$ 与叶节点开销 $n^{\log_b a} = 1$。两者同阶，满足 $f(n) = \Theta(n^{\log_b a})$。尽管问题的缩减因子为分数，但计算负载在递归层级间的分布依然是同质的。
* **结论：** 依据主定理 Case 2，结合递归树最大深度界限，得出系统整体复杂度。根据大 $O$ 符号的常数吸收特性，对数的底数（常数项）可被隐去。
  $$T(n) = \Theta(1 \cdot \log_{3/2} n) = \Theta(\log n)$$
* **渐近上界：** $O(\log n)$

### 4. $T(n) = 5T(n/4) + n$

* **分析：** 该递归模型表征了子问题生成率显著高于数据规模缩减率的系统。提取标准参数：$a=5, b=4, f(n)=n$。

* **比较：** 考察驱动项 $f(n) = n^1$ 与底端开销 $n^{\log_4 5}$。存在常数 $\epsilon \approx 0.161 > 0$，使得 $f(n) = O(n^{\log_b a - \epsilon})$ 成立。这证明底层叶节点引发的计算复杂性在多项式尺度上压倒了顶层的线性合并开销，系统呈现典型的“底端重载 (Bottom-heavy)”特征。
* **结论：** 依据主定理 Case 1（叶节点主导情形），总渐近运行时间完全受控于递归树的最底端叶节点规模。
  $$T(n) = \Theta(n^{\log_b a}) = \Theta(n^{\log_4 5})$$
* **渐近上界：** $O(n^{\log_4 5})$

### 5. $T(n) = T(\sqrt{n}) + 1$

* **分析：** 由于子问题规模呈现非线性衰减（指数级降维），本方程跳出了主定理的直接作用域。必须引入基于域映射的代换法 (Substitution Method) 进行同态转换。首先定义变量映射 $n = 2^m$，即 $m = \log_2 n$。将原方程 $T(n) = T(n^{1/2}) + 1$ 映射到新域可得 $T(2^m) = T(2^{m/2}) + 1$。引入辅助函数 $S(m) = T(2^m)$，方程被重构为标准的线性分治形式：$S(m) = S(m/2) + 1$。
* **计算：** 针对衍生方程 $S(m)$ 进行参数解析：$a=1, b=2, f(m)=1$。计算映射域中的叶节点状态量级：$m^{\log_b a} = m^{\log_2 1} = m^0 = 1$。
* **比较：** 在衍生域中，驱动项 $f(m) = 1$ 与界限 $m^{\log_2 1} = 1$ 严格同阶。完美契合主定理 Case 2 规范。
* **结论：** 首先推导得出衍生函数在 $m$ 域的紧确界：$S(m) = \Theta(m^{\log_b a} \log m) = \Theta(\log m)$。随后应用逆映射操作，将域参量 $m = \log_2 n$ 替换回原域，以求解初始函数的真实渐近行为。
  $$T(n) = S(m) = \Theta(\log m) = \Theta(\log(\log n))$$
* **渐近上界：** $O(\log \log n)$

### 2. 算法功能描述及其高效性判定

* **功能描述：** 该伪代码描述的是一个被称为 **Stooge Sort（斯图吉排序）** 的经典递归排序算法。其核心功能是**将输入数组 `A` 按照非降序（升序）进行排列**。
* **内在机制证明（归纳逻辑）：** 1. 当 $n=2$ 时，算法直接通过一次比较保证前两个元素有序。
  2. 当 $n \ge 3$ 时，算法采用“三分法”策略：首先递归排序数组的前 $2/3$ 部分；接着递归排序数组的后 $2/3$ 部分（这一步将整个数组中最大的前 $1/3$ 元素“挤”到了数组的最后端，确保末尾部分达到全局最大且有序）；最后，再次递归排序前 $2/3$ 部分（以保证剩余元素内部的相对有序性）。通过这种重叠覆盖的递归策略，最终实现全局有序。
* **高效性判定：** **该方案并非最高效的解决方案，因为存在比该算法更加高效的算法。**
* **次优性论证：**
  1. 理论下界对比：基于比较的排序算法，其信息论时间复杂度下界为 $\Omega(n \log n)$。经典的归并排序 (Merge Sort)、快速排序 (Quick Sort) 以及堆排序 (Heap Sort) 均能达到 $O(n \log n)$ 的最优渐近界。
  2. 极度低效的渐近表现：通过第一问的推导可知，本算法的复杂度高达 $O(n^{2.71})$。这表明其不仅远逊于上述 $O(n \log n)$ 的最优算法，甚至比诸如冒泡排序、插入排序这类被视为低效的 $O(n^2)$ 简单多项式时间算法还要缓慢得多。其主要学术价值在于作为复杂度分析与分治法设计失败案例的 pedagogical（教学化）探讨模型。

---

idea:

````python
# add your code here
def merge_sort(A: List[int]) -> List[int]:
    """
    高阶替代方案：Merge Sort，时间复杂度稳定在 O(n log n)。
    采用 out-of-place 方式构建，优化了合并阶段的指针调度。
    """
    # 递归终止条件
    if len(A) <= 1:
        return A
        
    # 均衡划分 (Divide)
    mid = len(A) // 2
    left_sub = merge_sort(A[:mid])
    right_sub = merge_sort(A[mid:])
    
    # 线性合并 (Conquer & Combine)
    return _merge(left_sub, right_sub)

def _merge(left: List[int], right: List[int]) -> List[int]:
    merged_array = []
    i = j = 0
    
    # 双指针游走合并
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged_array.append(left[i])
            i += 1
        else:
            merged_array.append(right[j])
            j += 1
            
    # 收集剩余尾部元素
    merged_array.extend(left[i:])
    merged_array.extend(right[j:])
    
    return merged_array
# your algorithm time complexity is:
# O(n log n)
````

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

### 理论模型构建与问题同构映射

在正式设计算法之前，需要对业务问题进行严格的数学抽象。题目给定一个收益数组 `profits`，其中 `profits[i]` 表示第 $i$ 天的收益波动。如果在第 $i$ 天买入，并在第 $j$ 天（$j \ge i$）卖出，由于数组记录的是“每日收益”（即当天的涨跌幅），我们在这一区间内的总利润恰好等于这段连续子数组的元素总和：
$$TotalProfit(i, j) = \sum_{k=i}^{j} profits[k]$$
因此，该问题在数学上被完美地等价映射为经典的**最大子数组和问题**。

---

### 1. 基础算法方案设计与分析

#### 1.1 算法分析与渐近复杂度
作为基线考察，最直观的暴力枚举法是遍历所有可能的买入日期 $i$ 和卖出日期 $j$ 的组合，计算区间内的收益和，并维护全局最大值。为了避免重复计算区间和，我们可以利用前缀和或在固定起点时进行累加。

* **时间复杂度：** 存在两层嵌套循环。外层循环确定买入点 $i$（$n$ 次），内层循环确定卖出点 $j$（平均 $n/2$ 次），总比较次数量级为 $n^2$。故渐近时间复杂度为 $O(n^2)$。
* **空间复杂度：** 仅需维护常数个标量变量（最大利润、买入点、卖出点等），不需要额外分配随输入规模增长的内存空间。故渐近空间复杂度为 $O(1)$。

idea:

````python
# add your code here
from typing import List, Tuple

def max_profit_baseline(profits: List[int]) -> Tuple[int, int, int]:
    """
    基础方案：时间复杂度 O(n^2) 的暴力求解法
    :param profits: 每日收益数组
    :return: (最大利润, 买入日期, 卖出日期)  # 日期采用 1-based 索引
    """
    n = len(profits)
    if n == 0:
        return 0, 0, 0
        
    max_profit = float('-inf')
    best_buy_day = -1
    best_sell_day = -1
    
    # 穷举所有可能的 (买入, 卖出) 区间组合
    for i in range(n):
        current_sum = 0
        for j in range(i, n):
            current_sum += profits[j]
            # 发生帕累托改进，更新最优解
            if current_sum > max_profit:
                max_profit = current_sum
                best_buy_day = i + 1  # 转换为现实中的第几天 (1-based)
                best_sell_day = j + 1
                
    return max_profit, best_buy_day, best_sell_day
# your algorithm time complexity is: 
# O(n^2)
````

### 1. $O(\log n)$ 时间复杂度下的单局部最小值检索

* **问题建模与理论可行性分析：**
  给定条件 $A[1] > A[2]$ 意味着序列在起始处具有负向的离散梯度（呈下降趋势），而 $A[n-1] < A[n]$ 表明序列在末端具有正向的离散梯度（呈上升趋势）。在一个元素互异的有限整数序列中，由连续性原理的离散模拟可知，必然在这两个边界之间存在至少一个极小值点（局部最小）。这种边界梯度符号相异的特性，为我们在离散空间中使用**二分查找 (Binary Search)** 提供了理论基础。

* **算法设计思路：**
  为了达到 $O(\log n)$ 的界，我们必须在每次迭代中丢弃一半的搜索空间。我们取区间中点 $mid$：
  1.  **极值判定：** 若 $A[mid] < A[mid-1]$ 且 $A[mid] < A[mid+1]$，则 $mid$ 即为所求的局部极小值，直接返回。
  2.  **空间剪枝（右偏）：** 若 $A[mid] > A[mid+1]$，说明 $mid$ 处的局部趋势是下降的。由于末端 $A[n]$ 处的趋势是上升的，因此在子区间 $[mid, n]$ 内必然存在由下降转为上升的拐点。我们可以安全地将搜索左边界向右推移至 $mid + 1$。
  3.  **空间剪枝（左偏）：** 若 $A[mid-1] < A[mid]$，说明 $mid$ 处的局部趋势是上升的。由于起始端 $A[1]$ 处的趋势是下降的，在子区间 $[1, mid]$ 内必然存在拐点。故将搜索右边界推移至 $mid - 1$。

* **复杂度计算：**
  每次判定消耗 $\Theta(1)$ 时间，搜索空间以 $1/2$ 的比例缩减。状态转移方程为 $T(n) = T(n/2) + \Theta(1)$。根据主定理，其渐近时间复杂度为严格的 $O(\log n)$；空间复杂度为 $O(1)$。

idea:

````python
# add your code here
from typing import List

def find_one_local_minimum(A: List[int]) -> int:
    """
    基于二分策略在 O(log n) 时间内检索单个局部最小值。
    假设输入数组 A 满足 len(A) >= 3, 且 A[0] > A[1], A[-2] < A[-1]。
    注: 算法基于 0-based 索引实现。
    """
    n = len(A)
    # 收敛搜索边界，排除必然不可能是局部最小的两个端点
    left, right = 1, n - 2
    
    while left <= right:
        mid = left + (right - left) // 2
        
        # 命中局部极小值条件
        if A[mid] < A[mid - 1] and A[mid] < A[mid + 1]:
            return A[mid]
            
        # 局部梯度下降，极小值必定存在于右侧子域
        elif A[mid] > A[mid + 1]:
            left = mid + 1
            
        # 局部梯度上升，极小值必定存在于左侧子域
        else:
            right = mid - 1
            
    # 理论上给定先决条件，循环内必然会 return，此处作防御性兜底
    return -1
# your algorithm time complexity is:
# O(log n)
````

### 2. 全局局部最小值集合检索与理论下界分析

* **计算复杂性下界：**
  当任务目标从“存在性检索（求单一极值）”放宽至“完备性检索（求所有极值）”时，问题的计算复杂性发生本质变迁。考虑最坏情况下的输入张量：一个呈现严格锯齿状震荡的序列（例如 $A = [10, 1, 9, 2, 8, 3, 7]$），其中局部极小值的分布密度可达 $O(n)$ 级别。由于任何声称能提取完备集合的算法，其算子都必须至少执行与结果集基数等同的 $O(n)$ 次内存写入操作，根据信息论下界原理，该任务存在严格的 $\Omega(n)$ 理论下界。这意味着，在渐近意义上，寻求亚线性复杂度的求解策略在拓扑上是不可能的。

* **算法架构与状态转移思路：**
  基于 $\Omega(n)$ 的绝对下界约束，该问题的最优逼近策略即为**单趟线性扫描 (Single-pass Linear Scan)**。
  根据题设给定的边界偏导条件 $A[1] > A[2]$ 与 $A[n-1] < A[n]$，序列的两端节点已被预先剥夺了成为局部最小值的拓扑可能性。因此，观测滑窗仅需在内部闭区间 $[2, n-1]$（对应 0-based 索引的 $[1, n-2]$）内平移。对于滑窗内的每个中心锚点 $i$，计算并验证其一阶差分状态（即 $A[i] < A[i-1] \land A[i] < A[i+1]$ ）。若特征匹配，则将其并入解集。

* **时空复杂度评估：**
  核心算子对目标数组进行了一次非中断的线性遍历，时间复杂度严格贴合理论最优界 $O(n)$。在内存开销方面，除了用于承载输出流的动态数组外，算法的辅助运行时状态仅依赖少数常数级寄存器，因此辅助空间复杂度为 $\Theta(1)$。

````python
from typing import List

def find_all_local_minima(A: List[int]) -> List[int]:
    """
    基于一维线性扫描的完备局部极小值检索算法。
    时间复杂度: O(n), 辅助空间复杂度: O(1)
    
    :param A: 满足边界条件 A[0]>A[1] 且 A[-2]<A[-1] 的无重复元素一维整型张量
    :return: 包含所有局部极小值的列表
    """
    n = len(A)
    
    if n < 3:
        return []
        
    local_minima_set = []
    
    # 滑窗规避首尾奇点，仅在内部拓扑空间进行邻域梯度判定
    for i in range(1, n - 1):
        # 谷底特征点捕获：要求双侧局部偏导数均为正
        if A[i] < A[i - 1] and A[i] < A[i + 1]:
            local_minima_set.append(A[i])
            
    return local_minima_set
````

### 1. 基于全局排序的 Top-$k$ 检索

* **算法选择与设计思路：**
    针对无序数组的最优渐近排序，**归并排序** 是一种基于分治范式的稳定排序算法。它在最坏、平均和最好情况下的时间复杂度均能严格保证为 $\Theta(n \log n)$。我们首先利用归并排序对全局数据进行非降序排列，随后通过切片操作直接提取前 $k$ 个元素。
* **渐近时间复杂度推导：**
    1.  **全局排序阶段：** 归并排序的递归动力学方程为 $T(n) = 2T(n/2) + \Theta(n)$。根据主定理 Case 2，该阶段的时间开销为 $\Theta(n \log n)$。
    2.  **截取输出阶段：** 在连续内存（如数组）中提取前 $k$ 个元素，仅需进行 $k$ 次内存读取与赋值，开销为 $\Theta(k)$。
    3.  **总体复杂度：** 综合两阶段，系统总时间开销为 $O(n \log n + k)$。由于约束条件中 $k \le n$，高阶项主导渐近行为，最终严格界定为 **$O(n \log n)$**。

idea:

````python
# add your code here
from typing import List

def merge_sort(arr: List[int]) -> List[int]:
    """归并排序的底层实现，保证 O(n log n) 的最坏情况复杂度"""
    if len(arr) <= 1:
        return arr
    
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    
    return _merge(left, right)

def _merge(left: List[int], right: List[int]) -> List[int]:
    merged = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1
    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged

def get_top_k_by_sorting(arr: List[int], k: int) -> List[int]:
    """
    策略一：全局排序后输出前 k 个极小值
    """
    if k <= 0 or not arr:
        return []
    sorted_arr = merge_sort(arr)
    return sorted_arr[:k]
# your algorithm time complexity is:
# O(n log n)
````

### 2. 基于最小堆连续插入的前 $k$ 个极小值检索

* **算法设计思路：**
    本策略采用优先队列（具体表现为二叉最小堆）来维护数据的拓扑偏序关系。根据题干约束，我们不采用时间开销更优的自底向上线性建堆法，而是严格执行**动态连续插入**：维护一个初始为空的堆，遍历输入数组，将元素逐个插入，并在堆内部通过上浮操作恢复堆的偏序性质。建堆完成后，连续执行 $k$ 次提取堆顶极小值及其伴随的下沉操作，依次按顺序输出结果。

* **时间复杂度严格证明：**
    1.  **建堆阶段：** 向规模为 $i-1$ 的堆中插入第 $i$ 个元素时，最坏情况下需要沿树高进行上浮操作，单次比较次数上限为 $\log_2 i$。遍历 $n$ 个元素构建全量堆的总时间开销为：
        $$W_{build}(n) = \sum_{i=1}^{n} O(\log i) = O(\log(n!))$$
        根据斯特林近似展开，可知 $\log(n!) = \Theta(n \log n)$。因此，这种通过往空堆中不断插入元素来建堆的方法，其时间复杂度严格界定为 **$O(n \log n)$**。
    2.  **提取阶段：** 每次取出堆顶极小值后，将底层尾部元素移至堆顶并触发下沉过程。单次下沉的开销取决于当前堆的最大深度，即 $O(\log n)$。连续执行 $k$ 次提取，该阶段累计开销为 **$O(k \log n)$**。
    3.  **总体复杂度：** 算法系统的整体开销由建堆与提取两部分构成，即 $O(n \log n + k \log n)$。考虑到约束条件中 $k \le n$ 必然成立，高阶项 $n \log n$ 占据主导地位，整体渐近时间复杂度上限最终收敛于 **$O(n \log n)$**。

````python
from typing import List

class MinHeap:
    """严格实现基于连续存储的一维数组最小堆"""
    def __init__(self):
        self.heap = []

    def insert(self, val: int):
        # 元素追加至尾部并执行上浮调整，单次最坏开销 O(log N)
        self.heap.append(val)
        self._bubble_up(len(self.heap) - 1)

    def extract_min(self) -> int:
        if not self.heap:
            raise IndexError("无法从空堆中提取元素")
        
        # 捕获堆顶极小值，并将尾部元素置换至堆顶后执行下沉调整
        min_val = self.heap[0]
        last_val = self.heap.pop()
        
        if self.heap:
            self.heap[0] = last_val
            self._bubble_down(0)
            
        return min_val

    def _bubble_up(self, index: int):
        parent = (index - 1) // 2
        # 当当前节点非根节点且严格小于其父节点时，持续上浮
        while index > 0 and self.heap[index] < self.heap[parent]:
            self.heap[index], self.heap[parent] = self.heap[parent], self.heap[index]
            index = parent
            parent = (index - 1) // 2

    def _bubble_down(self, index: int):
        size = len(self.heap)
        while True:
            smallest = index
            left = 2 * index + 1
            right = 2 * index + 2

            # 在当前节点与其所有有效子节点中寻找最小值的索引
            if left < size and self.heap[left] < self.heap[smallest]:
                smallest = left
            if right < size and self.heap[right] < self.heap[smallest]:
                smallest = right

            # 若满足最小堆性质则终止；否则交换并继续向下递归
            if smallest != index:
                self.heap[index], self.heap[smallest] = self.heap[smallest], self.heap[index]
                index = smallest
            else:
                break

def get_top_k_by_heap_insertion(arr: List[int], k: int) -> List[int]:
    """
    通过往空堆中不断插入元素的方法建堆，并顺序输出最小的 k 个数
    """
    if k <= 0 or not arr:
        return []
        
    min_heap = MinHeap()
    
    # 阶段 1: 连续插入法建堆
    for num in arr:
        min_heap.insert(num)
        
    # 阶段 2: 提取前 k 个最小元素
    result = []
    extract_count = min(k, len(arr))
    for _ in range(extract_count):
        result.append(min_heap.extract_min())
        
    return result
````

### 星球碰撞问题的算法设计与解析

* **问题建模与拓扑结构分析：**
  本问题描述了一个在一维线性空间中发生的序列化状态决议过程。由于所有星球的速度绝对值相同，碰撞**仅当**一个向右运动的星球（状态为正）位于一个向左运动的星球（状态为负）的左侧时才会发生。这种“最近的相邻异向元素优先作用”的特性，在拓扑结构上完美契合**栈** 数据结构的“后进先出 (LIFO)”范式。

* **算法设计思路 (基于栈的模拟机制)：**
  构建一个工作栈来动态维护存活的星球序列，并对输入数组进行单向线性扫描：
  1.  **右向运动入栈：** 若当前遍历到的星球 $p > 0$（自左向右运动），它不会与栈中已有的任何星球发生碰撞（因为栈中的星球要么与它同向等速，要么背向而去）。因此，直接将其压入栈中。
  2.  **左向运动决议：** 若当前星球 $p < 0$（自右向左运动），它将不可避免地与栈顶的向右运动星球（$> 0$）发生碰撞。此时触发碰撞结算循环：
      * 比较质量大小（即绝对值）。
      * 若 $|p| > \text{栈顶质量}$：栈顶的向右运动星球被摧毁（出栈），继续利用当前星球 $p$ 与新的栈顶进行下一轮碰撞测算。
      * 若 $|p| < \text{栈顶质量}$：当前星球 $p$ 被摧毁，终止碰撞结算（不入栈）。
      * 若 $|p| == \text{栈顶质量}$：根据题设特殊的边界条件（“自右向左运动的星球消失，自左向右的星球保持不变”），当前向左运动的星球 $p$ 湮灭，而栈顶的向右星球存活。此时同样终止碰撞结算。
  3.  **幸存者入栈：** 若碰撞结算循环自然结束（栈被清空，或栈顶元素也为向左运动的负数），说明当前向左运动的星球摧毁了沿途所有阻碍，成功存活，将其压入栈中。

* **时空复杂性分析 (基于摊还分析)：**
  * **时间复杂度：** 尽管在单次遍历的 `for` 循环内部嵌套了一个 `while` 循环，但依据**摊还分析 (Amortized Analysis)**，输入数组中的每一个星球最多被压入栈一次，且最多被弹出栈一次。由于栈的 `push` 和 `pop` 操作均在 $\Theta(1)$ 时间内完成，因此整体状态机流转的指令总数上限严格受控于 $2n$。系统的渐近时间复杂度为严格的 **$O(n)$**。
  * **空间复杂度：** 在最坏情况下（例如所有星球均向右运动，或者所有星球均向左运动，无任何碰撞发生），所有元素都将被压入并驻留在栈中。因此，辅助空间复杂度为 **$O(n)$**。

idea:

````python
# add your code here
from typing import List

def simulate_planet_collisions(A: List[int]) -> List[int]:
    """
    基于栈范式模拟一维空间内的星球碰撞过程。
    时间复杂度: O(n), 空间复杂度: O(n)
    """
    stack = []
    
    for p in A:
        if p > 0:
            # 向右运动的星球，暂时安全，压入存活栈
            stack.append(p)
        else:
            # 向左运动的星球，触发与向右运动星球的碰撞决议
            survived = True
            
            # 当栈非空，且栈顶星球向右运动时，必然发生碰撞
            while stack and stack[-1] > 0:
                # 提取质量(绝对值)进行对抗比较
                mass_p = abs(p)
                mass_stack_top = stack[-1]
                
                if mass_p > mass_stack_top:
                    # 栈顶星球被摧毁，出栈，p 继续向前撞击
                    stack.pop()
                elif mass_p < mass_stack_top:
                    # 当前星球 p 被摧毁，停止撞击
                    survived = False
                    break
                else:
                    # 质量相等：根据题设"自右向左的星球消失，自左向右的保持不变"
                    # 当前星球 p (向左) 消失，栈顶星球 (向右) 存活
                    survived = False
                    break
            
            # 若当前向左运动的星球未被摧毁（清空了前方的正数，或前方本就安全）
            if survived:
                stack.append(p)
                
    return stack
# your algorithm time complexity is:
# O(n)
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

### 0-1 背包问题的算法设计与解析

* **问题建模：**
  本问题属于经典的组合优化问题。给定大理石集合的规模 $N = 4$，小推车的最大承载量 $W = 9$。每个大理石具有两个属性：重量 $w_i$ 与价值 $v_i$。约束条件为选取的元素集合的总重量不得超过 $W$，目标是最大化这些元素的总价值。由于每块大理石仅具备“放入”或“不放入”两种离散状态，且不可分割，因此该问题严格映射为 0-1 背包模型。

* **算法设计思路：**
  解决此类最优子结构问题，**动态规划**是理论上最为完备的策略。
  1.  **状态定义：** 设 $f(i, j)$ 表示在前 $i$ 块大理石中进行选择，且总重量不超过 $j$ 的情况下，所能获取的最大价值。
  2.  **状态转移方程：** 对于第 $i$ 块大理石，存在两种决策逻辑：
      * 若当前剩余容量不足以容纳该大理石（即 $j < w_i$），则状态继承自上一层：
        $$f(i, j) = f(i-1, j)$$
      * 若容量充足（即 $j \ge w_i$），则需要在“不放入该大理石”与“放入该大理石”两者之间取最优解：
        $$f(i, j) = \max\{f(i-1, j), f(i-1, j-w_i) + v_i\}$$
  3.  **空间状态压缩：** 观察状态转移方程可知，第 $i$ 层的状态计算仅依赖于第 $i-1$ 层的状态。为了优化内存消耗，可将二维状态矩阵压缩为一维数组 $dp[j]$。在迭代计算时，容量维度 $j$ 必须采取逆序遍历（从 $W$ 递减至 $w_i$），以此确保在更新 $dp[j]$ 时，引用的 $dp[j-w_i]$ 依然是上一层（即未放入第 $i$ 块大理石）的状态，从而严格遵守 0-1 背包中每个物品仅限使用一次的约束。

* **时空复杂度评估：**
  * **时间复杂度：** 算法包含两层嵌套循环，外层遍历 $N$ 个物品，内层遍历最大容量 $W$。状态转移的核心计算为常数时间操作，故渐近时间复杂度为 $O(N \cdot W)$。在本例中计算量极小，属于伪多项式时间复杂度。
  * **空间复杂度：** 采用一维数组进行状态压缩后，仅需维护一个长度为 $W + 1$ 的数组。故辅助空间复杂度为 $O(W)$。


idea:

````python
# add your code here
from typing import List, Tuple

def solve_01_knapsack(items: List[Tuple[int, int]], capacity: int) -> int:
    """
    基于状态压缩的 0-1 背包求解算法。
    时间复杂度: O(N*W), 空间复杂度: O(W)
    
    :param items: 大理石属性列表，元素格式为 (重量, 价值)
    :param capacity: 小推车最大承重
    :return: 最大可装载的价值
    """
    # 初始化一维状态数组，dp[j] 表示容量为 j 时的最大价值
    dp = [0] * (capacity + 1)
    
    for weight, value in items:
        # 逆序遍历背包容量，防止物品状态被重复计算
        for j in range(capacity, weight - 1, -1):
            dp[j] = max(dp[j], dp[j - weight] + value)
            
    return dp[capacity]
# your algorithm time complexity is:
# O(N*W)
````

### 恢复二叉搜索树的算法设计与解析

* **问题建模与拓扑特征分析：**
  二叉搜索树 (Binary Search Tree, BST) 的核心拓扑不变量在于：其**中序遍历** 序列严格单调递增。题设指出树中有且仅有两个节点的值发生错误交换，这在代数表现上等价于在一个严格递增序列中交换了两个元素，从而必然破坏序列的单调性，产生局部的**逆序对**。
  设原递增序列中被交换的两个元素为 $x$ 和 $y$（且 $x < y$），交换后序列出现两种形态：
  1.  **相邻交换：** 仅产生一个逆序对。例如序列 `[1, 2, 4, 3, 5]` 中，逆序对为 `(4, 3)`。交换的节点即为该逆序对的两个元素。
  2.  **非相邻交换：** 产生两个逆序对。例如序列 `[1, 5, 3, 4, 2, 6]` 中，逆序对为 `(5, 3)` 和 `(4, 2)`。交换的节点分别为第一个逆序对的首元素（$5$）与第二个逆序对的尾元素（$2$）。

* **算法设计思路：**
  基于上述分析，算法转化为在中序遍历序列中捕获逆序对的过程。
  1.  维护一个 `prev` 指针，记录中序遍历序列中的前驱节点。
  2.  在深度优先搜索 (DFS) 过程中，实时比对 `prev.val` 与当前节点 `curr.val`。若满足 `prev.val > curr.val`，则触发逆序异常。
  3.  利用两个全局状态指针 `first` 和 `second` 锁定目标：
      * 首次捕获异常时，将 `prev` 赋值给 `first`，`curr` 赋值给 `second`（兼容相邻交换的情形）。
      * 若后续再次捕获异常，仅更新 `second` 为当前的 `curr`（处理非相邻交换的情形）。
  4.  遍历终止后，对 `first.val` 与 `second.val` 执行标量交换，即可在不破坏树体物理结构的条件下恢复全局偏序性质。

* **时空复杂度评估：**
  * **时间复杂度：** 算法需对 BST 进行一次完整的中序遍历以准确定位故障节点，状态比较为常数时间操作。因此渐近时间复杂度严格为 $O(N)$，其中 $N$ 为树中节点总数。
  * **空间复杂度：** 采用递归方式进行 DFS，其调用栈的深度等价于树的最大深度 $H$。最坏情况下（树退化为链表）空间复杂度为 $O(N)$；平均情况下（树结构相对平衡）空间复杂度为 $O(\log N)$。整体渐近界记为 $O(H)$。（注：若严格要求 $O(1)$ 辅助空间，可采用 Morris 中序遍历进行拓扑线索化替换，但本实现以标准 DFS 为例以保证结构清晰）。

idea:

````python
# add your code here
from typing import Optional

# 节点结构定义
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def recoverTree(self, root: Optional[TreeNode]) -> None:
        """
        基于中序遍历的二叉搜索树异常节点恢复算法。
        要求：原地修改 (In-place modification)
        """
        self.first_node = None
        self.second_node = None
        # 初始化前驱节点为负无穷，保证首次比较的鲁棒性
        self.prev_node = TreeNode(float('-inf'))

        def inorder_traverse(node: TreeNode):
            if not node:
                return
            
            # 左子树遍历
            inorder_traverse(node.left)
            
            # 拓扑异常检测：违反单调递增公理
            if self.prev_node.val > node.val:
                # 捕获首个逆序对的首元素
                if not self.first_node:
                    self.first_node = self.prev_node
                # 滚动更新尾元素（兼容单/双逆序对情形）
                self.second_node = node
                
            # 状态演进
            self.prev_node = node
            
            # 右子树遍历
            inorder_traverse(node.right)

        inorder_traverse(root)
        
        # 物理置换异常节点的值，完成状态自愈
        if self.first_node and self.second_node:
            self.first_node.val, self.second_node.val = self.second_node.val, self.first_node.val
# your algorithm time complexity is:
# O(n)
````

### 接雨水问题的算法设计与解析

* **问题建模：**
  给定一个一维非负整数数组 $height$，表示地形的高度图。对于数组中的任意局部位置 $i$，其上方能够蓄水的体积 $V_i$ 严格受限于所谓的“木桶效应”，即由其左侧的最高点 $L\_max_i$ 与右侧的最高点 $R\_max_i$ 中的较小值决定。
  在代数表示上，局部蓄水量满足公式：
  $$V_i = \max(0, \min(L\_max_i, R\_max_i) - height[i])$$
  全局总蓄水量即为 $\sum V_i$。

* **算法设计思路（双指针策略）：**
  最朴素的动态规划法需要预先正向和逆向遍历数组，以 $O(N)$ 的空间复杂度建立完整的 $L\_max$ 和 $R\_max$ 备忘录。为实现空间复杂度的极小化，可采用**双指针法 (Two Pointers)** 动态收敛搜索空间。
  
  1.  **初始化：** 设置左右指针 `left = 0` 和 `right = n - 1`，并维护当前左侧与右侧的已知最高水位边界 `left_max = 0`, `right_max = 0`。
  2.  **动态收敛条件：** 当 `left < right` 时，交替推进两端指针。
      * 每次迭代先更新局部极值：`left_max = max(left_max, height[left])`，`right_max = max(right_max, height[right])`。
      * **核心决议逻辑：** 若 $height[left] < height[right]$，根据单调性可知，$R\_max_{left}$ 必然大于等于 $height[right]$。因此，决定 `left` 处蓄水量的瓶颈完全且唯一地依赖于 `left_max`。此时可直接结算 `left` 处的蓄水量并推进左指针。
      * 反之，若 $height[left] \ge height[right]$，同理可知 `right` 处的蓄水量瓶颈由 `right_max` 决定。结算 `right` 处蓄水量并推进右指针。

* **时空复杂度评估：**
  * **时间复杂度：** 左右指针交替向中间收敛，数组中的每个元素仅被访问并计算一次，状态转移的常数极小。故渐近时间复杂度为严格的 $O(N)$。
  * **空间复杂度：** 算法运行期间仅维护了常数个标量寄存器（如指针索引、累加器及边界极值），未申请任何随输入规模线性增长的动态内存。故辅助空间复杂度为 $O(1)$，达到理论下界。

idea:

````python
# add your code here
from typing import List

def trap_rain_water(height: List[int]) -> int:
    """
    基于双指针策略的接雨水算法。
    时间复杂度: O(N), 空间复杂度: O(1)
    
    :param height: 一维非负整数数组，表征柱形高度图
    :return: 能够捕获的雨水总量
    """
    n = len(height)
    if n < 3:
        # 拓扑退化防御：不足以形成谷底结构的序列无法蓄水
        return 0

    left, right = 0, n - 1
    left_max, right_max = 0, 0
    total_water = 0

    while left < right:
        # 边界极值状态更新
        left_max = max(left_max, height[left])
        right_max = max(right_max, height[right])

        # 依据木桶效应判定局部蓄水量瓶颈
        if height[left] < height[right]:
            # 左侧边界成为约束条件
            total_water += left_max - height[left]
            left += 1
        else:
            # 右侧边界成为约束条件
            total_water += right_max - height[right]
            right -= 1

    return total_water
# your algorithm time complexity is:
# O(N)
````

### 根到叶节点数字之和的算法设计与解析

* **问题建模与代数特征分析：**
  给定一棵节点值域定义在 $[0, 9]$ 闭区间的二叉树。从根节点出发至任意叶节点的有向拓扑路径唯一确定一个十进制整数。
  设节点 $v$ 在拓扑深度 $d$ 处，其承载的值为 $val(v)$。若以自顶向下的视角考察，父节点向子节点传递数值时，等价于在十进制向量空间中进行左移并执行标量加法。即状态转移方程可抽象为：
  $$S(v) = S(parent) \times 10 + val(v)$$
  其中 $S(v)$ 表征从根节点到当前节点 $v$ 累积生成的十进制数值。算法的核心目标是求解所有满足“出度为0”（即叶节点）的节点 $u$ 的 $S(u)$ 之总和。

* **算法设计思路 (基于深度优先搜索 DFS)：**
  针对此类依赖于完整路径状态聚合的树形拓扑问题，**前序遍历 (Pre-order Traversal)** 是理论上最为契合的搜索范式。
  1.  **状态传递：** 设计一个递归子程序，维护当前遍历路径的隐式状态变量 `current_sum`。每当访问一个新节点时，利用上述状态转移方程实时更新 `current_sum`。
  2.  **边界收敛与决议：** * 若当前节点为空，即触碰物理边界，对总和的贡献度为 $0$，予以剪枝返回。
      * 若当前节点为叶节点（左右子指针均为空），说明一条完整的十进制数字序列构建完毕，直接返回当前所持有的 `current_sum` 状态值。
  3.  **结果分治聚合：** 对于非叶节点，其所在路径产生的最终数字和，严格等于其左子树衍生路径产生的数字和加上右子树衍生路径产生的数字和。通过递归回溯进行标量累加即可。

* **时空复杂性评估：**
  * **时间复杂度：** 算法通过深度优先策略对二叉树进行了一次完备的图遍历，树中的每个节点被且仅被访问一次。节点内部的状态计算与条件分支耗时均为 $\Theta(1)$。因此，系统的渐近时间复杂度严格收敛于 $O(N)$，其中 $N$ 为树体节点总数。
  * **空间复杂度：** 算法未显式开辟多项式级别的辅助数据结构。内存开销完全取决于操作系统隐式维护的递归调用栈。栈的最大深度等于树的拓扑高度 $H$。在最佳情况（完全二叉树）下为 $O(\log N)$；在最坏情况（退化为单向链表）下界为 $O(N)$。整体渐近辅助空间复杂度记为 $O(H)$。

idea:

````python
# add your code here
from typing import Optional

# 节点拓扑结构定义
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def sumNumbers(self, root: Optional[TreeNode]) -> int:
        """
        基于前序 DFS 范式的根到叶节点路径数值求和算法。
        时间复杂度: O(N), 空间复杂度: O(H)
        """
        def dfs(node: Optional[TreeNode], current_sum: int) -> int:
            # 基础边界：空节点无数值贡献
            if not node:
                return 0
            
            # 状态转移：十进制位左移与当前节点值累加
            current_sum = current_sum * 10 + node.val
            
            # 拓扑极点检测：抵达叶节点，闭合路径并返回状态聚合值
            if not node.left and not node.right:
                return current_sum
                
            # 分治与回溯：聚合左右子拓扑空间的路径总和
            return dfs(node.left, current_sum) + dfs(node.right, current_sum)
            
        return dfs(root, 0)
# your algorithm time complexity is:
# O(N)  
````
